```python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-
"""
================================================================================
JOUR 6 : NetworkX — Visualisation de base avec Matplotlib
Résumé complet et commenté
================================================================================
Ce script reprend l'ensemble des concepts vus au Jour 6.
"""

import networkx as nx
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
from matplotlib.patches import Patch
import numpy as np

# ==============================================================================
# 6.1 DESSIN DE BASE
# ==============================================================================
print("=" * 70)
print("6.1 DESSIN DE BASE")
print("=" * 70)

G = nx.karate_club_graph()
```

```python
# --- nx.draw() — fonction générique ---
fig, ax = plt.subplots(figsize=(10, 8))
nx.draw(G, with_labels=True, node_color='lightblue', node_size=500,
        font_size=8, font_weight='bold', edge_color='gray', ax=ax)
ax.set_title('nx.draw() — Dessin rapide et complet')
plt.close()
print("Figure 01_draw_base affichée")
```

```python
# --- draw_networkx() — ne nettoie pas l'axe ---
fig, ax = plt.subplots(figsize=(10, 8))
pos = nx.spring_layout(G, seed=42)
# On peut superposer : d'abord le fond, puis les nœuds en rouge
nx.draw_networkx_edges(G, pos=pos, alpha=0.3, ax=ax)
nx.draw_networkx_nodes(G, pos=pos,
                       nodelist=[0, 1, 2, 3], node_color='red', node_size=800, ax=ax)
nx.draw_networkx_labels(G, pos=pos, font_size=8, ax=ax)
ax.set_title('draw_networkx() — Superposition manuelle (nœuds 0-3 en rouge)')
plt.close()
print("Figure 02_draw_networkx affichée")
```

```python
# --- draw_networkx_nodes() — contrôle fin des nœuds ---
fig, ax = plt.subplots(figsize=(10, 8))
pos = nx.spring_layout(G, seed=42)
# Tous les nœuds en gris clair
nx.draw_networkx_nodes(G, pos, node_color='lightgray', node_size=400, ax=ax)
# Sous-ensemble en rouge avec contour noir
nx.draw_networkx_nodes(G, pos, nodelist=[0, 33], node_color='crimson',
                       node_size=1000, edgecolors='black', linewidths=2.5, ax=ax)
nx.draw_networkx_edges(G, pos, alpha=0.4, ax=ax)
nx.draw_networkx_labels(G, pos, font_size=8, ax=ax)
ax.set_title('draw_networkx_nodes() — Sous-ensemble et contours')
plt.close()
print("Figure 03_draw_nodes affichée")
```

```python
# --- draw_networkx_edges() — contrôle fin des arêtes ---
fig, ax = plt.subplots(figsize=(10, 8))
pos = nx.spring_layout(G, seed=42)
nx.draw_networkx_nodes(G, pos, node_color='skyblue', node_size=500, ax=ax)
# Arêtes spécifiques en rouge, épaisses, en pointillés
highlighted_edges = [(0, 1), (0, 2), (0, 3)]
nx.draw_networkx_edges(G, pos, edgelist=highlighted_edges,
                       width=3, edge_color='red', style='dashed', ax=ax)
# Le reste en gris fin
remaining_edges = [edge for edge in G.edges() if edge not in highlighted_edges]
nx.draw_networkx_edges(G, pos, edgelist=remaining_edges,
                       width=1, edge_color='gray', alpha=0.5, ax=ax)
nx.draw_networkx_labels(G, pos, font_size=8, ax=ax)
ax.set_title('draw_networkx_edges() — Sous-ensemble, style et couleur')
plt.close()
print("Figure 04_draw_edges affichée")
```

```python
# --- draw_networkx_labels() — labels personnalisés ---
fig, ax = plt.subplots(figsize=(10, 8))
pos = nx.spring_layout(G, seed=42)
nx.draw_networkx_nodes(G, pos, node_color='navy', node_size=600, ax=ax)
nx.draw_networkx_edges(G, pos, alpha=0.3, ax=ax)
# Labels personnalisés avec fond coloré
labels = {node: f"N{node}" for node in G.nodes()}
nx.draw_networkx_labels(G, pos, labels=labels, font_size=10, font_color='white',
                        font_weight='bold',
                        bbox=dict(boxstyle='round,pad=0.3', facecolor='darkblue', alpha=0.9), ax=ax)
ax.set_title('draw_networkx_labels() — Labels personnalisés avec bbox')
plt.close()
print("Figure 05_draw_labels affichée")
```

```python
# --- draw_networkx_edge_labels() — poids sur les arêtes ---
fig, ax = plt.subplots(figsize=(10, 8))
G_weighted = nx.Graph()
G_weighted.add_weighted_edges_from([
    (1, 2, 5), (1, 3, 2), (2, 3, 7), (2, 4, 3), (3, 4, 1), (4, 5, 8)
])
pos_w = nx.spring_layout(G_weighted, seed=42, k=2)
nx.draw_networkx_nodes(G_weighted, pos_w, node_color='lightgreen',
                       node_size=1200, edgecolors='black', ax=ax)
nx.draw_networkx_edges(G_weighted, pos_w, width=2, edge_color='darkgreen', alpha=0.7, ax=ax)
nx.draw_networkx_labels(G_weighted, pos_w, font_size=12, font_weight='bold', ax=ax)
# Labels de poids
edge_labels = {(u, v): data['weight'] for u, v, data in G_weighted.edges(data=True)}
nx.draw_networkx_edge_labels(G_weighted, pos_w, edge_labels=edge_labels,
                             font_size=11, label_pos=0.5,
                             bbox=dict(boxstyle='round,pad=0.2', facecolor='yellow', alpha=0.9), ax=ax)
ax.set_title('draw_networkx_edge_labels() — Poids sur les arêtes')
plt.close()
print("Figure 06_draw_edge_labels affichée")
```
